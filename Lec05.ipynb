```python
weekly_sales = {}

while True:
    item = input().strip().lower()

    if item == "end":
        break

    quantity = int(input(f"How many {item}s were sold?: "))

    if item in weekly_sales:
        weekly_sales[item] += quantity
    else:
        weekly_sales[item] = quantity
    print(f"Recorded: +{quantity} {item}(s)\n")

    for item, total_sold in weekly_sales.items():
        print(f"• {item.capitalize()}: {total_sold} sold")
```

```python
import numpy as np

mat = np.random.randint(1, 91, size=(5, 5))
print("matrix :" , mat)

row_wise_means = np.mean(mat, axis=1)
column_wise_means = np.mean(mat, axis=0)

print("Row-wise means:", row_wise_means)
print("Column-wise means:", column_wise_means)

modified_matrix = np.where(mat % 2 == 0, -1, mat)
print("Matrix with even numbers replaced by -1:")
print(modified_matrix)

sorted_row = np.sort(mat, axis=1)
sorted_column = np.sort(mat, axis=0)

print("Row-wise sorted matrix:")
print(sorted_row)
print("Column-wise sorted matrix:")
print(sorted_column)

rowiwse_sum = np.sum(mat, axis=1)
column_wise_sum = np.sum(mat, axis=0)

print("Row-wise sum:", rowiwse_sum)
print("Column-wise sum:", column_wise_sum)

count_col = np.sum(np.where(mat < 0, 1, 0), axis=0)
print(count_col)
```
